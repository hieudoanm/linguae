# WiLI 2018 Text-to-CSV

Converts the WiLI 2018 `x_train.txt`, `y_train.txt`, `x_test.txt`, and `y_test.txt`
files into a single `data/csv/wili-2018.csv` with columns `text` and `language`.

```python
from pathlib import Path

import pandas as pd

BASE = Path(__file__).resolve().parent.parent
DATA_DIR = BASE / "data" / "txt" / "wili-2018"
OUT = BASE / "data" / "csv" / "wili-2018.csv"

x_train = (DATA_DIR / "x_train.txt").read_text(encoding="utf-8").splitlines()
y_train = (DATA_DIR / "y_train.txt").read_text(encoding="utf-8").splitlines()
x_test = (DATA_DIR / "x_test.txt").read_text(encoding="utf-8").splitlines()
y_test = (DATA_DIR / "y_test.txt").read_text(encoding="utf-8").splitlines()

assert len(x_train) == len(y_train), (len(x_train), len(y_train))
assert len(x_test) == len(y_test), (len(x_test), len(y_test))

df_train = pd.DataFrame({"text": x_train, "language": y_train})
df_test = pd.DataFrame({"text": x_test, "language": y_test})
df = pd.concat([df_train, df_test], ignore_index=True)

print(f"train: {len(df_train)}  test: {len(df_test)}  total: {len(df)}")
print(df["language"].value_counts().head())
```

```python
OUT.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(OUT, index=False, encoding="utf-8")
print(f"Wrote {OUT} ({len(df)} rows)")
```
